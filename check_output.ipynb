```python
import numpy as np
import matplotlib.pyplot as plt
import os

from configs.config_object import ConfigObject
```

```python
config = ConfigObject('config.json')
base_output_folder = f'./99-output'
```

```python
output_folder = os.path.join(base_output_folder, config.simulation_name)
binary_data_folder = os.path.join(output_folder, 'binary_data')
```

## BC bin file

```python
bin_bc_south_v_path = os.path.join(binary_data_folder, 'bc_south_v.bin')
```

```python
with open(bin_bc_south_v_path, 'rb') as fid:
    bin_bc_south_v = np.fromfile(fid, dtype='>f8')
```

```python
nx = config.Nx
nz = len(bin_bc_south_v) / nx
```

```python
arr_bc_south_v = np.reshape(bin_bc_south_v, (int(nz),int(nx)))
```
